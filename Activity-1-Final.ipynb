# Case Study 1: Customer Review Text Analysis

## Objective

The objective of this case study is to preprocess customer reviews using the NLTK library.

The analysis includes:

- Sentence tokenization
- Word tokenization
- Counting sentences in each review
- Counting words in each sentence
- Identifying positive and negative words
- Flagging reviews containing more than 3 sentences

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
reviews = [
    "The product quality is excellent! I am very satisfied with the purchase. Will definitely recommend it.",
    
    "Terrible customer service. I had to wait two weeks for a response. Not buying again!",
    
    "Average product. Delivery was on time, but the packaging was damaged."
]
```

```python
positive_words = [
    "excellent",
    "satisfied",
    "recommend"
]

negative_words = [
    "terrible",
    "damaged"
]
```

```python
for review_number, review in enumerate(reviews, start=1):

    print("=" * 60)
    print("REVIEW", review_number)
    print("=" * 60)

    # Sentence tokenization
    sentences = sent_tokenize(review)

    print("\nNumber of sentences:", len(sentences))

    # Process each sentence
    for sentence_number, sentence in enumerate(sentences, start=1):

        print("\nSentence", sentence_number, ":", sentence)

        # Word tokenization
        words = word_tokenize(sentence)

        # Count actual words, excluding punctuation
        word_count = 0

        for word in words:
            if word.isalpha():
                word_count += 1

        print("Words:", words)
        print("Number of words:", word_count)

        # Identify positive and negative words
        for word in words:
            clean_word = word.lower()

            if clean_word in positive_words:
                print("Positive word:", clean_word)

            elif clean_word in negative_words:
                print("Negative word:", clean_word)

    # Flag reviews with more than 3 sentences
    if len(sentences) > 3:
        print("\nFLAGGED: More than 3 sentences")
    else:
        print("\nNot flagged")
```

```python
def flag_review(review):
    sentences = sent_tokenize(review)
    sentence_count = len(sentences)

    if sentence_count > 3:
        return "FLAGGED"
    else:
        return "Not flagged"
```

```python
for review_number, review in enumerate(reviews, start=1):
    print("Review", review_number, ":", flag_review(review))
```

## Conclusion

The customer reviews were successfully tokenized into sentences and words using NLTK.

The analysis identified positive and negative words from the reviews and calculated the number of sentences and words in each sentence.

A review-flagging function was also implemented to identify reviews containing more than 3 sentences for further analysis.

This preprocessing provides a foundation for more advanced Natural Language Processing (NLP) and sentiment analysis tasks.
